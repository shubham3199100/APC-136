```python
# Program 1: Shape Area - Runtime Polymorphism
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return 0.5 * self.base * self.height

shapes = [Circle(5), Rectangle(10, 5), Triangle(10, 6)]
for s in shapes:
    print("Area:", s.area())
```

```python
# Program 2: Employee Salary - Method Overriding
class Employee:
    def calculate_salary(self):
        pass

class Manager(Employee):
    def calculate_salary(self):
        basic = 50000
        return basic + basic * 0.30

class Developer(Employee):
    def calculate_salary(self):
        basic = 40000
        return basic + basic * 0.20

class Tester(Employee):
    def calculate_salary(self):
        basic = 35000
        return basic + basic * 0.15

employees = [Manager(), Developer(), Tester()]
for e in employees:
    print("Salary:", e.calculate_salary())
```

```python
# Program 3: Vehicle Start - Method Overriding
class Vehicle:
    def start(self):
        print("Vehicle starts")

class Car(Vehicle):
    def start(self):
        print("Car starts with a key/button")

class Bike(Vehicle):
    def start(self):
        print("Bike starts with self-start")

class Bus(Vehicle):
    def start(self):
        print("Bus starts with ignition")

vehicles = [Car(), Bike(), Bus()]
for v in vehicles:
    v.start()
```

```python
# Program 9: Distance Addition - Operator Overloading
class Distance:
    def __init__(self, feet, inches):
        self.feet = feet
        self.inches = inches

    def __add__(self, other):
        feet = self.feet + other.feet
        inches = self.inches + other.inches
        feet += inches // 12
        inches = inches % 12
        return Distance(feet, inches)

    def display(self):
        print(self.feet, "feet", self.inches, "inches")

d1 = Distance(5, 8)
d2 = Distance(3, 7)
d3 = d1 + d2
d3.display()
```

```python
# Program 10: Student Marks Comparison - Operator Overloading
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks

    def __lt__(self, other):
        return self.marks < other.marks

s1 = Student("Amit", 80)
s2 = Student("Rahul", 75)

print("s1 > s2:", s1 > s2)
print("s1 < s2:", s1 < s2)
```

```python
# Program 11: Product Price Comparison - Operator Overloading
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __eq__(self, other):
        return self.price == other.price

    def __gt__(self, other):
        return self.price > other.price

p1 = Product("Laptop", 50000)
p2 = Product("Phone", 30000)

print("Equal Price:", p1 == p2)
print("Product 1 Costlier:", p1 > p2)
```

```python
# Program 12: Online Shopping Payment - Polymorphism
class Payment:
    def make_payment(self, amount):
        pass

class UPIPayment(Payment):
    def make_payment(self, amount):
        print("Paid", amount, "using UPI")

class CardPayment(Payment):
    def make_payment(self, amount):
        print("Paid", amount, "using Card")

class WalletPayment(Payment):
    def make_payment(self, amount):
        print("Paid", amount, "using Wallet")

def pay(payment, amount):
    payment.make_payment(amount)

pay(UPIPayment(), 500)
pay(CardPayment(), 1000)
pay(WalletPayment(), 300)
```

```python
# Program 15: Smart Device - Method Overriding
class SmartDevice:
    def turn_on(self):
        pass
    def turn_off(self):
        pass

class Light(SmartDevice):
    def turn_on(self):
        print("Light turned ON")
    def turn_off(self):
        print("Light turned OFF")

class Fan(SmartDevice):
    def turn_on(self):
        print("Fan turned ON")
    def turn_off(self):
        print("Fan turned OFF")

class AC(SmartDevice):
    def turn_on(self):
        print("AC turned ON")
    def turn_off(self):
        print("AC turned OFF")

class TV(SmartDevice):
    def turn_on(self):
        print("TV turned ON")
    def turn_off(self):
        print("TV turned OFF")

devices = [Light(), Fan(), AC(), TV()]
for d in devices:
    d.turn_on()
    d.turn_off()
```
